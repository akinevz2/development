```python
# lab_week4.py
from gensim.models import KeyedVectors
import json

# Load the model (this might take a while)
filename = "GoogleNews-vectors-negative300.bin"
print("Loading model...")
mymodel = KeyedVectors.load_word2vec_format(filename, binary=True)
print("Model loaded!")

# Test basic functionality
print("Similarity between 'man' and 'woman':", mymodel.similarity('man', 'woman'))
print("Most similar to 'man':", mymodel.most_similar(positive=['man']))
print("Vector for 'man':", mymodel['man'])

# Test analogy
print("Analogy result:", mymodel.most_similar(positive=['China', 'London'], negative=['England']))

# Load relations
with open('relations.json', 'r') as fp:
    testtuples = json.load(fp)
print("Relations loaded:", testtuples)
```
